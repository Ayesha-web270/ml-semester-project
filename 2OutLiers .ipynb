```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r'C:\Users\ST\Downloads\1Sleep_health_and_lifestyle_dataset.csv')
```

```python
df.head()
```

```python
# Box Plot of Age
sns.boxplot(df['Age'])
```

```python
# Outliers detection of Age
percentile25=df['Age'].quantile(0.25)
percentile75=df['Age'].quantile(0.75)
iqr=percentile75-percentile25
ub=percentile75+1.5*iqr
lb=percentile25-1.5*iqr
print('Upper bound is ',ub)
print('Upper bound is ',lb)
```

```python
# Capping Age
df['Age']=np.where(
    df['Age']>ub,
    ub,
np.where(
   df['Age']<lb,
    lb,
    df['Age']
)    
)
```

```python
sns.boxplot(df['Age'])
```
