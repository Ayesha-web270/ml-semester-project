```python
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(drop='first', sparse_output=False, handle_unknown='ignore')
X_train_ohe = ohe.fit_transform(X_train[['Gender','Occupation']]).astype('int64')
X_test_ohe = ohe.transform(X_test[['Gender','Occupation']]).astype('int64')
```

```python
X_train_ohe
```

```python
train_nominal_encoded_df = pd.DataFrame(X_train_ohe, 
                                        columns=ohe.get_feature_names_out(), 
                                        index=X_train.index)
test_nominal_encoded_df = pd.DataFrame(X_test_ohe, 
                                       columns=ohe.get_feature_names_out(), 
                                       index=X_test.index)
train_nominal_encoded_df
```

```python

X_train = X_train.drop(columns=['Gender','Occupation']).join(train_nominal_encoded_df)
X_test = X_test.drop(columns=['Gender','Occupation']).join(test_nominal_encoded_df)

X_train.shape
```

```python
#for multiple ordinal feature
from sklearn.preprocessing import OrdinalEncoder
ordinal_orders = [['Overweight','Normal','Normal Weight', 'Obese']]
oe = OrdinalEncoder(categories=ordinal_orders)
X_train[['BMI Category']] = oe.fit_transform(X_train[['BMI Category']])
X_test[['BMI Category']] = oe.transform(X_test[['BMI Category']])
```

```python
train_ordinal_encoded_df = pd.DataFrame(X_train_ohe,
                                        columns=ohe.get_feature_names_out(),
                                        index=X_train.index)
test_ordinal_encoded_df = pd.DataFrame(X_test_ohe,
                                      columns=ohe.get_feature_names_out(),
                                        index=X_test.index) 
train_ordinal_encoded_df
```
